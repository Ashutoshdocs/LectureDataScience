# Step 1 · Remove extra spaces & tabs
**Panama Tech Academy** · Input: `raw_data.txt` → Output: `outputs/step1_stripped.txt`

```python
with open("raw_data.txt", encoding="utf-8") as f:
    lines = f.read().split("\n")             # split the text into lines (removes the \n)

print(lines[:6])                             # BEFORE: see the spaces and \t
```

```python
lines = [line.strip() for line in lines]    # strip every line
print(lines[:6])                             # AFTER
```

```python
with open("outputs/step1_stripped.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(lines))                # join the lines back with \n and save
print("Saved outputs/step1_stripped.txt")
```
