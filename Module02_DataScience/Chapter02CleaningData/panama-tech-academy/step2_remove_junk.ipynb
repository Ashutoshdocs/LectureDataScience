# Step 2 · Remove junk lines
**Panama Tech Academy** · Input: `outputs/step1_stripped.txt` → Output: `outputs/step2_no_junk.txt`

```python
with open("outputs/step1_stripped.txt", encoding="utf-8") as f:
    lines = f.read().split("\n")             # read the output of step 1

junk = {"Follow", "Message"}                 # Instagram button text copied by mistake
print("Junk lines found:", sum(1 for l in lines if l in junk))   # count them first
```

```python
lines = [line for line in lines if line not in junk]   # keep everything that is not junk

with open("outputs/step2_no_junk.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(lines))
print("Saved outputs/step2_no_junk.txt")
```
