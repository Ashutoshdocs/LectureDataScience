# Step 3 · Fix count words (POSTS → posts)
**Panama Tech Academy** · Input: `outputs/step2_no_junk.txt` → Output: `outputs/step3_fixed_words.txt`

```python
with open("outputs/step2_no_junk.txt", encoding="utf-8") as f:
    lines = f.read().split("\n")

count_words = ("post", "posts", "follower", "followers", "following")

fixed = []
for line in lines:
    parts = line.rsplit(" ", 1)              # split off the last word: ["18.6K", "Followers"]
    if len(parts) == 2 and parts[1].lower() in count_words:
        new_line = parts[0] + " " + parts[1].lower()  # lowercase only the word, keep "18.6K"
        if new_line != line:                 # only show lines that actually changed
            print(f"{line!r:24} -> {new_line}")
        line = new_line
    fixed.append(line)
```

```python
with open("outputs/step3_fixed_words.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(fixed))
print("Saved outputs/step3_fixed_words.txt")
```
