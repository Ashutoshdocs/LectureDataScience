# Step 6 · Handle missing values
**Panama Tech Academy** · Input: `outputs/step5_emails_booleans.csv` → Output: `outputs/step6_no_missing.csv`

```python
import pandas as pd                     # pandas = tables (DataFrames)
import numpy as np                      # numpy gives us np.nan (a missing value)

df = pd.read_csv("outputs/step5_emails_booleans.csv", parse_dates=["join_date"])
df.isna().sum()                              # BEFORE
```

```python
df["age"] = df["age"].fillna(df["age"].median())   # missing age -> middle age of everyone

df["salary"] = (df.groupby("department")["salary"]       # group rows by department
                .transform(lambda s: s.fillna(s.median())))  # fill with that department's median
df["salary"] = df["salary"].fillna(df["salary"].median())  # rows with no department -> overall median

df["city"] = df["city"].fillna("Unknown")              # text columns -> "Unknown"
df["department"] = df["department"].fillna("Unknown")
df["is_active"] = df["is_active"].fillna(False).astype(bool)   # unknown status -> not active

df.isna().sum()                              # AFTER: only join_date and email stay missing (can't guess them)
```

```python
df.to_csv("outputs/step6_no_missing.csv", index=False)
print("Saved outputs/step6_no_missing.csv")
```
