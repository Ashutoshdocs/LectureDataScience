# Step 4 · Parse mixed date formats
**Panama Tech Academy** · Input: `outputs/step3_fixed_numbers.csv` → Output: `outputs/step4_dates.csv`

```python
import pandas as pd                     # pandas = tables (DataFrames)
import numpy as np                      # numpy gives us np.nan (a missing value)

df = pd.read_csv("outputs/step3_fixed_numbers.csv")
print(df["join_date"].unique())              # BEFORE: 4 different formats
```

```python
formats = ("%Y-%m-%d",                       # 2023-01-15
           "%d/%m/%Y",                       # 15/02/2023
           "%Y.%m.%d",                       # 2023.03.20
           "%B %d %Y")                       # March 5 2023

def parse_date(x):
    if pd.isna(x):                           # empty cell -> no date
        return pd.NaT                        # NaT = "Not a Time" (missing date)
    for fmt in formats:                      # try each format one by one
        try:
            return pd.to_datetime(x, format=fmt)   # success -> return the date
        except ValueError:                   # wrong format -> try the next one
            continue
    return pd.NaT                            # nothing matched

df["join_date"] = df["join_date"].apply(parse_date)   # apply() runs the function on every cell
df["join_date"].value_counts(dropna=False)   # AFTER: one format, plus NaT for blanks
```

```python
df.to_csv("outputs/step4_dates.csv", index=False)   # dates are saved as 2023-01-15
print("Saved outputs/step4_dates.csv")
```
