# Step 7 · Structure & export clean data
**Panama Tech Academy** · Input: `outputs/step6_no_missing.csv` → Output: `outputs/clean_data.csv` + `outputs/clean_data.json`

```python
import pandas as pd                     # pandas = tables (DataFrames)
import numpy as np                      # numpy gives us np.nan (a missing value)

df = pd.read_csv("outputs/step6_no_missing.csv", parse_dates=["join_date"])
```

```python
df = df.astype({"id": "int64",               # whole numbers
                "age": "int64",
                "salary": "float64",          # decimal numbers
                "is_active": "bool",          # True / False
                "city": "category",           # a small set of repeated values
                "department": "category"})

df["salary_band"] = pd.cut(df["salary"],                        # put salaries into buckets
                           bins=[0, 50000, 100000, np.inf],      # 0-50k, 50k-100k, 100k+
                           labels=["Low", "Medium", "High"])
df["join_year"] = df["join_date"].dt.year.astype("Int64")       # year from the date (Int64 allows missing)

df = df[["id", "name", "age", "city", "department", "salary", "salary_band",   # nice column order
         "join_date", "join_year", "email", "is_active"]]
df = df.sort_values("id").reset_index(drop=True)                 # sort by id

df.info()
df.head(10)
```

```python
df.to_csv("outputs/clean_data.csv", index=False)                 # clean table as CSV
df.to_json("outputs/clean_data.json", orient="records",           # one JSON object per row
           indent=4, date_format="iso")                           # dates as 2023-01-15T00:00:00
print("Saved clean_data.csv and clean_data.json", df.shape)
```
