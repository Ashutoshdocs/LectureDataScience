# Step 8 · Summarise by department
**Panama Tech Academy** · Input: `outputs/clean_data.csv` → Output: `outputs/department_summary.csv`

```python
import pandas as pd

df = pd.read_csv("outputs/clean_data.csv")   # the clean data from step 7
```

```python
summary = (df.groupby("department")                 # one row per department
             .agg(employees=("id", "count"),          # how many people
                  avg_age=("age", "mean"),            # average age
                  avg_salary=("salary", "mean"),      # average salary
                  active=("is_active", "sum"))        # True counts as 1, so sum = number active
             .round(1))                               # 1 decimal place
summary
```

```python
summary.to_csv("outputs/department_summary.csv")   # keep the index (department names)
print("Saved outputs/department_summary.csv")
```
