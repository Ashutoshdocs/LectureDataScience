# Step 3 · Fix numeric columns (age, salary)
**Panama Tech Academy** · Input: `outputs/step2_clean_text.csv` → Output: `outputs/step3_fixed_numbers.csv`

```python
import pandas as pd                     # pandas = tables (DataFrames)
import numpy as np                      # numpy gives us np.nan (a missing value)

df = pd.read_csv("outputs/step2_clean_text.csv")
print(df["age"].unique())                    # BEFORE: "thirty", "-5", "150" ...
```

```python
df["age"] = pd.to_numeric(df["age"], errors="coerce")   # text that isn't a number ("thirty") -> NaN
df.loc[~df["age"].between(18, 70), "age"] = np.nan        # ages outside 18-70 are impossible -> NaN
```

```python
df["salary"] = (df["salary"].astype(str)              # make sure every value is text
                .str.replace(r"[₹,\s]", "", regex=True)) # remove ₹, commas and spaces: "₹45,000" -> "45000"
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")   # text -> number ("nan" text -> NaN)
df.loc[df["salary"] > 500000, "salary"] = np.nan         # 9999999 is an unrealistic outlier -> NaN

df[["age", "salary"]].describe()             # AFTER: min / max now look sensible
```

```python
df.to_csv("outputs/step3_fixed_numbers.csv", index=False)
print("Saved outputs/step3_fixed_numbers.csv")
```
