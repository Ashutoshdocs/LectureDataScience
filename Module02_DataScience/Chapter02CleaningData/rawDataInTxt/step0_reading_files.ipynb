# Step 0 · Reading & writing files
**Panama Tech Academy** · Input: `raw_data.txt` → Output: `outputs/step0_report.txt`

### 1. Open, read, close

```python
f = open("raw_data.txt", encoding="utf-8")   # open the file for reading (default mode is "r")
text = f.read()                              # read() gives the WHOLE file as one string
f.close()                                    # always close a file when you are done

print(type(text))                            # it is a str
print(len(text), "characters")               # how many characters in the file
```

### 2. The better way: `with` (closes the file for you)

```python
with open("raw_data.txt", encoding="utf-8") as f:   # file closes automatically after this block
    text = f.read()

print(text[:120])                            # print the first 120 characters
```

### 3. Read one line at a time

```python
with open("raw_data.txt", encoding="utf-8") as f:
    first = f.readline()                     # readline() reads ONE line
    second = f.readline()                    # the next call reads the next line

print(repr(first))                           # repr shows hidden \n and spaces
print(repr(second))
```

### 4. Read all lines into a list

```python
with open("raw_data.txt", encoding="utf-8") as f:
    lines = f.readlines()                    # readlines() gives a list, one item per line

print(len(lines), "lines")
print(lines[:5])                             # first 5 lines (notice the \n at the end of each)
```

### 5. Loop over a file line by line

```python
count = 0
with open("raw_data.txt", encoding="utf-8") as f:
    for line in f:                           # a file can be looped directly, one line each time
        if line.strip() == "":               # strip() removes spaces/\n; empty means blank line
            count += 1                       # count blank lines
print("Blank lines:", count)
```

### 6. Write an output file

```python
import os
os.makedirs("outputs", exist_ok=True)        # create the outputs folder (no error if it exists)

with open("outputs/step0_report.txt", "w", encoding="utf-8") as f:   # "w" = write (overwrites)
    f.write("Raw data report\n")              # write() does NOT add a newline, so we add \n
    f.write(f"Characters: {len(text)}\n")
    f.write(f"Lines: {len(lines)}\n")
    f.write(f"Blank lines: {count}\n")

with open("outputs/step0_report.txt", "a", encoding="utf-8") as f:   # "a" = append (adds to the end)
    f.write("First line: " + lines[0].strip() + "\n")

print(open("outputs/step0_report.txt", encoding="utf-8").read())    # read it back to check
```
