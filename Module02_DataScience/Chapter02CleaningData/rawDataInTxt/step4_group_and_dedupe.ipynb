# Step 4 · Group into profiles & remove duplicates
**Panama Tech Academy** · Input: `outputs/step3_fixed_words.txt` → Output: `outputs/step4_cleaned.txt`

```python
with open("outputs/step3_fixed_words.txt", encoding="utf-8") as f:
    lines = f.read().split("\n")

profiles, current = [], []                   # all profiles / the one being built
for line in lines:
    if line == "":                           # blank line = end of a profile
        if current:
            profiles.append(current)
        current = []
    else:
        current.append(line)
if current:                                  # last profile (no blank line after it)
    profiles.append(current)
print("Profiles found:", len(profiles))
```

```python
seen, unique = set(), []
for p in profiles:
    if p[0] in seen:                         # p[0] is the username
        print("Duplicate removed:", p[0])
        continue
    seen.add(p[0])
    unique.append(p)
print("Unique profiles:", len(unique))
```

```python
with open("outputs/step4_cleaned.txt", "w", encoding="utf-8") as f:
    f.write("\n\n".join("\n".join(p) for p in unique) + "\n")   # one blank line between profiles
print("Saved outputs/step4_cleaned.txt")
```
