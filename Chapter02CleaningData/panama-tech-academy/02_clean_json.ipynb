# 02 · Organize into clean JSON
**Panama Tech Academy** · Input: `cleaned_data.txt` → Output: `clean_data.json`

```python
with open("cleaned_data.txt", encoding="utf-8") as f:
    text = f.read()                     # read the cleaned text

chunks = text.strip().split("\n\n")    # after cleaning, ONE blank line separates profiles
print("Profiles:", len(chunks))
```

### Turn `2,300` / `18.6K` / `1.2M` into numbers

```python
def to_number(value):
    value = value.replace(",", "")      # "2,300" -> "2300"
    if value.endswith("K"):             # thousands
        return int(float(value[:-1]) * 1_000)       # "18.6K" -> 18600
    if value.endswith("M"):             # millions
        return int(float(value[:-1]) * 1_000_000)   # "1.2M" -> 1200000
    return int(value)                   # plain number

print(to_number("2,300"), to_number("18.6K"), to_number("1.2M"))   # quick check
```

### Turn one profile into a dictionary

```python
def parse_profile(chunk):
    lines = chunk.split("\n")           # one item per line
    return {
        "username": lines[0],                           # line 1: username
        "posts": to_number(lines[1].split(" ")[0]),     # line 2: "1,240 posts" -> 1240
        "followers": to_number(lines[2].split(" ")[0]), # line 3: "18.6K followers" -> 18600
        "following": to_number(lines[3].split(" ")[0]), # line 4: "210 following" -> 210
        "name": lines[4],                               # line 5: display name
        "category": lines[5] if len(lines) > 5 else "Unknown",   # line 6 if it exists
        "bio": "\n".join(lines[6:]),                    # everything after is the bio
    }

parse_profile(chunks[0])                # test on the first profile
```

### Parse all profiles and save JSON

```python
import json                             # built-in module for JSON

data = [parse_profile(c) for c in chunks]           # parse every profile into a list of dicts

with open("clean_data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, indent=4, ensure_ascii=False)  # indent=4 = readable, ensure_ascii=False keeps emojis

print("Saved", len(data), "profiles to clean_data.json")
```
