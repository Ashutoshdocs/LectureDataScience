**Panama Tech Academy** · Input: `clean_data.json`

```python
import json

with open("clean_data.json", encoding="utf-8") as f:
    data = json.load(f)                 # list of profile dictionaries

print("Profiles loaded:", len(data))
```

### Who has the maximum posts?

```python
top = data[0]                           # start by assuming the first profile is the biggest
for p in data:                          # check every profile
    if p["posts"] > top["posts"]:       # found someone with more posts?
        top = p                         # they become the new top
print(top["username"], "-", top["posts"], "posts")
```

### Who has the maximum followers?

```python
top = data[0]
for p in data:
    if p["followers"] > top["followers"]:   # same idea, compare followers
        top = p
print(top["username"], "-", top["followers"], "followers")
```

### Who follows the maximum people?

```python
top = data[0]
for p in data:
    if p["following"] > top["following"]:   # compare following
        top = p
print(top["username"], "-", top["following"], "following")
```

### How many categories, and how many people in each?

```python
counts = {}                             # category -> number of people
for p in data:
    cat = p["category"]
    counts[cat] = counts.get(cat, 0) + 1    # get() returns 0 the first time we see a category

print("Categories:", len(counts))        # number of different categories
for cat, n in sorted(counts.items(), key=lambda x: x[1], reverse=True):   # biggest first
    print(f"  {cat:<24}{n}")
```

### Shortcut: the same answers with `max()`

```python
print(max(data, key=lambda p: p["posts"])["username"])       # key tells max() what to compare
print(max(data, key=lambda p: p["followers"])["username"])
print(max(data, key=lambda p: p["following"])["username"])
```
