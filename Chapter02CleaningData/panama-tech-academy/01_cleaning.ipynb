# 01 · Cleaning the raw data
**Panama Tech Academy** · Input: `raw_data.txt` → Output: `cleaned_data.txt`

```python
# open the raw file (utf-8 because bios contain emojis)
with open("raw_data.txt", encoding="utf-8") as f:
    raw = f.read()                      # read the whole file into one string

print(repr(raw[:300]))                  # repr() shows hidden spaces, tabs and \n so we can see the mess
```

### Remove extra spaces and tabs

```python
lines = raw.split("\n")                 # break the text into a list of lines
lines = [line.strip() for line in lines] # strip() removes spaces/tabs at both ends of every line
print(lines[:12])                       # peek at the first few cleaned lines
```

### Remove the Instagram buttons that were copied by mistake

```python
junk = {"Follow", "Message"}            # button text that is not real data
lines = [line for line in lines if line not in junk]   # keep only lines that are not junk
```

### Make the count words lowercase (`POSTS` → `posts`)

```python
count_words = ("post", "posts", "follower", "followers", "following")   # words that follow a number

fixed = []                              # new list for the fixed lines
for line in lines:                      # go through every line
    parts = line.rsplit(" ", 1)         # split off the LAST word, e.g. ["1,240", "Posts"]
    if len(parts) == 2 and parts[1].lower() in count_words:   # is the last word a count word?
        line = parts[0] + " " + parts[1].lower()              # lowercase only the word, not "18.6K"
    fixed.append(line)                  # save the (maybe changed) line
lines = fixed                           # replace the old list
```

### Group lines into profiles (blank line = new profile)

```python
profiles = []                           # list of profiles, each profile is a list of lines
current = []                            # lines of the profile we are building right now
for line in lines:
    if line == "":                      # blank line means the profile ended
        if current:                     # only save if we actually collected lines
            profiles.append(current)
        current = []                    # start a fresh profile
    else:
        current.append(line)            # normal line: add it to the current profile
if current:                             # file may not end with a blank line
    profiles.append(current)

print("Profiles found:", len(profiles))
```

### Remove duplicate profiles

```python
seen = set()                            # usernames we have already kept
unique_profiles = []
for profile in profiles:
    username = profile[0]               # first line of every profile is the username
    if username in seen:                # already kept this person?
        print("Duplicate removed:", username)
        continue                        # skip it
    seen.add(username)                  # remember the username
    unique_profiles.append(profile)     # keep the profile

print("Unique profiles:", len(unique_profiles))
```

### Save the cleaned file

```python
blocks = ["\n".join(p) for p in unique_profiles]   # join each profile's lines back into text
cleaned = "\n\n".join(blocks) + "\n"                # exactly ONE blank line between profiles

with open("cleaned_data.txt", "w", encoding="utf-8") as f:
    f.write(cleaned)                    # write the clean text to a new file

print(cleaned[:250])                    # show the start of the cleaned file
```
