# Step 6 · Answer the questions
**Panama Tech Academy** · Input: `outputs/step5_clean_data.json` → Output: `outputs/step6_answers.txt`

```python
import json
with open("outputs/step5_clean_data.json", encoding="utf-8") as f:
    data = json.load(f)                      # back to a list of dictionaries
```

```python
most_posts = max(data, key=lambda p: p["posts"])          # profile with most posts
most_followers = max(data, key=lambda p: p["followers"])  # profile with most followers
most_following = max(data, key=lambda p: p["following"])  # profile following the most people

counts = {}
for p in data:
    counts[p["category"]] = counts.get(p["category"], 0) + 1   # people per category
```

```python
with open("outputs/step6_answers.txt", "w", encoding="utf-8") as f:
    f.write(f"Maximum posts:     {most_posts['username']} ({most_posts['posts']})\n")
    f.write(f"Maximum followers: {most_followers['username']} ({most_followers['followers']})\n")
    f.write(f"Follows the most:  {most_following['username']} ({most_following['following']})\n")
    f.write(f"Categories:        {len(counts)}\n\n")
    for cat, n in sorted(counts.items(), key=lambda x: x[1], reverse=True):
        f.write(f"  {cat:<24}{n}\n")

print(open("outputs/step6_answers.txt", encoding="utf-8").read())
```
