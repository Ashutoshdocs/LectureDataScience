# Step 5 · Organize into JSON
**Panama Tech Academy** · Input: `outputs/step4_cleaned.txt` → Output: `outputs/step5_clean_data.json`

```python
def to_number(value):
    value = value.replace(",", "")           # "2,300" -> "2300"
    if value.endswith("K"):
        return int(float(value[:-1]) * 1_000)        # "18.6K" -> 18600
    if value.endswith("M"):
        return int(float(value[:-1]) * 1_000_000)    # "1.2M"  -> 1200000
    return int(value)


def parse_profile(chunk):
    lines = chunk.split("\n")
    return {
        "username": lines[0],
        "posts": to_number(lines[1].split(" ")[0]),       # "1,240 posts" -> 1240
        "followers": to_number(lines[2].split(" ")[0]),
        "following": to_number(lines[3].split(" ")[0]),
        "name": lines[4],
        "category": lines[5] if len(lines) > 5 else "Unknown",   # missing category
        "bio": "\n".join(lines[6:]),
    }
```

```python
import json

with open("outputs/step4_cleaned.txt", encoding="utf-8") as f:
    chunks = f.read().strip().split("\n\n")   # one profile per chunk

data = [parse_profile(c) for c in chunks]

with open("outputs/step5_clean_data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, indent=4, ensure_ascii=False)   # ensure_ascii=False keeps emojis readable

print("Saved", len(data), "profiles")
data[0]
```
