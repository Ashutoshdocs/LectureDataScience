# Step 1 · Remove duplicate rows
**Panama Tech Academy** · Input: `raw_data.csv` → Output: `outputs/step1_no_duplicates.csv`

```python
import pandas as pd                     # pandas = tables (DataFrames)
import numpy as np                      # numpy gives us np.nan (a missing value)

df = pd.read_csv("raw_data.csv")             # start from the raw file
before = len(df)                             # row count before cleaning

df = df.drop_duplicates()                    # remove rows that are exact copies
df = df.reset_index(drop=True)               # renumber rows 0, 1, 2 ... again

print(f"Removed {before - len(df)} duplicates -> {len(df)} rows")
df.to_csv("outputs/step1_no_duplicates.csv", index=False)   # index=False = don't save row numbers
```
