# Step 0 · Reading a CSV file & inspecting it
**Panama Tech Academy** · Input: `raw_data.csv` → Output: `outputs/step0_report.txt`

### 1. Read the file as plain text first

```python
with open("raw_data.csv", encoding="utf-8") as f:   # open the CSV like any text file
    lines = f.readlines()                    # list of lines

print(len(lines), "lines (1 header + data rows)")
for line in lines[:4]:                       # first 4 lines
    print(repr(line))                        # repr shows hidden spaces and \n
```

### 2. Read it with pandas

```python
import pandas as pd                     # pandas = tables (DataFrames)
import numpy as np                      # numpy gives us np.nan (a missing value)
pd.set_option("display.max_columns", None)   # show every column when printing

df = pd.read_csv("raw_data.csv")             # read the CSV into a DataFrame (a table)
df.head(10)                                  # show the first 10 rows
```

### 3. Inspect

```python
print("Shape:", df.shape)                    # (rows, columns)
df.info()                                    # column names, types and non-empty counts
```

```python
df.isna().sum()                              # missing values per column
```

```python
print("Duplicate rows:", df.duplicated().sum())    # rows that are exact copies
for col in ["city", "department", "is_active"]:    # columns with messy text
    print(f"\n{col}:", df[col].unique())          # every different value in that column
```

### 4. Save a report

```python
import os
os.makedirs("outputs", exist_ok=True)        # create the outputs folder

with open("outputs/step0_report.txt", "w", encoding="utf-8") as f:   # "w" = write
    f.write(f"Rows, columns: {df.shape}\n")
    f.write(f"Duplicate rows: {df.duplicated().sum()}\n\n")
    f.write("Missing values per column:\n")
    f.write(df.isna().sum().to_string())    # to_string() turns the result into text

print(open("outputs/step0_report.txt", encoding="utf-8").read())
```
