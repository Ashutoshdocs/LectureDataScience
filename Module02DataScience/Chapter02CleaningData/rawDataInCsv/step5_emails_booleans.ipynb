# Step 5 · Validate emails & fix yes/no values
**Panama Tech Academy** · Input: `outputs/step4_dates.csv` → Output: `outputs/step5_emails_booleans.csv`

```python
import pandas as pd                     # pandas = tables (DataFrames)
import numpy as np                      # numpy gives us np.nan (a missing value)

df = pd.read_csv("outputs/step4_dates.csv", parse_dates=["join_date"])   # parse_dates reads dates as dates
```

```python
pattern = r"^[\w.+-]+@[\w-]+\.[\w.]+$"         # name @ domain . ending
valid = df["email"].str.match(pattern, na=False)   # True if the email matches (blank -> False)
print("Invalid emails:", df.loc[~valid & df["email"].notna(), "email"].tolist())
df.loc[~valid, "email"] = np.nan             # invalid emails -> missing
```

```python
print(df["is_active"].unique())              # BEFORE: Y, yes, 1, N, no, 0, NaN

bool_map = {"y": True, "yes": True, "1": True,     # every way of saying yes
            "n": False, "no": False, "0": False}   # every way of saying no
df["is_active"] = (df["is_active"].astype(str)     # make everything text ("1", "Y" ...)
                   .str.strip().str.lower()        # "Y" -> "y"
                   .map(bool_map))                 # look up True/False (unknown -> NaN)

print(df["is_active"].unique())              # AFTER
```

```python
df.to_csv("outputs/step5_emails_booleans.csv", index=False)
print("Saved outputs/step5_emails_booleans.csv")
```
