# Step 2 · Clean text columns
**Panama Tech Academy** · Input: `outputs/step1_no_duplicates.csv` → Output: `outputs/step2_clean_text.csv`

```python
import pandas as pd                     # pandas = tables (DataFrames)
import numpy as np                      # numpy gives us np.nan (a missing value)

df = pd.read_csv("outputs/step1_no_duplicates.csv")   # output of step 1
df["city"].unique()                          # BEFORE: pune, PUNE , Bengaluru, New Delhi ...
```

```python
df["name"] = df["name"].str.strip().str.title()     # " rohan mehta" -> "Rohan Mehta"

df["department"] = (df["department"].str.strip()    # remove spaces: "engineering " -> "engineering"
                    .str.title()                     # "engineering" -> "Engineering"
                    .replace({"Hr": "HR"}))          # title() makes "Hr", fix it back to "HR"

city_map = {"bengaluru": "Bangalore", "new delhi": "Delhi"}   # different spellings -> one name
df["city"] = (df["city"].str.strip()        # remove spaces: "PUNE " -> "PUNE"
              .str.lower()                  # "PUNE" -> "pune"
              .replace(city_map)            # "bengaluru" -> "Bangalore"
              .str.title())                 # "pune" -> "Pune"

df["city"].unique()                          # AFTER
```

```python
df.to_csv("outputs/step2_clean_text.csv", index=False)
print("Saved outputs/step2_clean_text.csv")
```
